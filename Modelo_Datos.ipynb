### Modelo 1: Datos Basales

```python
TipoTransaccion = Enum("TipoTransaccion", {e: e for e in df_config.tipo_transaccion.dropna().unique()})

class Fase1Registro(BaseModel):
    tipo_transaccion: TipoTransaccion = Field(description=descr["tipo_transaccion"])
    numero_cbr_actual: int = Field(description=descr["numero_cbr_actual"])
    fojas_cbr_actual: str = Field(pattern = r'\b\d+\s*(?:bis|v)?\b', description=descr["fojas_cbr_actual"])
    dia_cbr_actual: int = Field(ge=1, le=31, description=descr["dia_cbr_actual"])
    mes_cbr_actual: int = Field(ge=1, le=12, description=descr["mes_cbr_actual"])
    ano_cbr_actual: int = Field(ge=1900, le=2035, description=descr["ano_cbr_actual"])
```

### Modelo 2: Caracterización Derecho

```python
class TipoDerecho(str, Enum):
    CONSUNTIVO = "CONSUNTIVO"
    NO_CONSUNTIVO = "NO CONSUNTIVO"
    NO_INDICA = "NO INDICA/OTRO"
    
class NaturalezaAgua(str, Enum):
    SUPERFICIAL_CORRIENTE = "SUPERFICIAL Y CORRIENTE"
    SUPERFICIAL_DETENIDA = "SUPERFICIAL Y DETENIDA"
    SUPERFICIAL = "SUPERFICIAL CORRIENTES/DETENIDAS"
    SUBTERRANEA = "SUBTERRÁNEA"
    NO_INDICA = "NO INDICA/OTRO"

class EjercicioDerecho(str, Enum):
    PERMANENTE_CONTINUO = "PERMANENTE Y CONTINUO"
    PERMANENTE_ALTERNADO = "PERMANENTE Y ALTERNADO"
    PERMANENTE_DISCONTINUO = "PERMANENTE Y DISCONTINUO"
    PERMANENTE_NO_ESPECIFICADO = "PERMANENTE Y NO ESPECIFICADO"
    EVENTUAL_CONTINUO = "EVENTUAL Y CONTINUO"
    EVENTUAL_DISCONTINUO = "EVENTUAL Y DISCONTINUO"
    EVENTUAL_ALTERNADO = "EVENTUAL Y ALTERNADO"
    EVENTUAL_NO_ESPECIFICADO = "EVENTUAL Y NO ESPECIFICADO"
    NO_INDICA = "NO INDICA/OTRO"

class UsoAgua(str, Enum):
    AGRICULTURA = "AGRICULTURA"
    MINERIA = "MINERÍA"
    INDUSTRIA = "INDUSTRIA"
    INDUSTRIAL = "INDUSTRIAL"
    GANADERIA = "GANADERÍA"
    ACUICULTURA = "ACUICULTURA"
    FORESTAL = "FORESTAL"
    AGUA_POTABLE = "AGUA POTABLE"
    USO_DOMESTICO = "USO DOMÉSTICO"
    SANITARIO = "SANITARIO"
    GENERACION_ELECTRICA = "GENERACIÓN ELÉCTRICA"
    RECARGA_ACUIFERO = "RECARGA DE ACUÍFERO"
    NO_INDICA = "NO INDICA/OTRO"

class Fase2Registro(BaseModel):
    tipo_derecho: List[TipoDerecho] = Field(description=descr["tipo_derecho"])
    naturaleza_agua: List[NaturalezaAgua] = Field(description=descr["naturaleza_agua"])
    ejercicio_derecho: List[EjercicioDerecho] = Field(description=descr["ejercicio_derecho"])
    uso_agua: List[UsoAgua] = Field(description=descr["uso_agua"])
```

### Modelo 3: Personas

```python
class Persona(str, Enum):
    EMISOR = "EMISOR"
    RECEPTOR = "RECEPTOR"

class Emisor(str, Enum):
    EMISOR = "EMISOR"

RutType = Annotated[str, Field(pattern=r'^(?:\d{1,2}\.\d{3}\.\d{3}-[\dK]|NO INDICA/OTRO)$')]

class Fase3Registro(BaseModel):
    tipo_persona: List[Persona] = Field(description=descr["tipo_persona"])
    rut: List[RutType] = Field(description=descr["rut"])
    nombre: List[str] = Field(description=descr["nombre"])
    nombre_representante_legal: Optional[List[str]] = Field(None,  description=descr["nombre_representante_legal"])
    rut_representante_legal: Optional[List[RutType]] = Field(None, description=descr["rut_representante_legal"])
    direccion: Optional[List[str]] = Field(None, description=descr["direccion"])
```

### Modelo 4: Caudal

```python
class TipoFuentes(str, Enum):
    ARROYO = "ARROYO"
    CANAL = "CANAL"
    CHARCA_AGUADA = "CHARCA / AGUADA"
    CHORRILLO = "CHORRILLO"
    DERRAME = "DERRAME"
    DREN = "DREN"
    EMBALSE_TRANQUE = "EMBALSE / TRANQUE"
    GALERIA = "GALERÍA / SOCAVÓN"
    LAGO_LAGUNA = "LAGO / LAGUNA"
    NORIA = "NORIA"
    POZO = "POZO (PROFUNDO)"
    PUNTERA = "PUNTERA"
    QUEBRADA = "QUEBRADA"
    RIO_ESTERO = "RÍO / ESTERO"
    VERTIENTE = "VERTIENTE"
    ZANJON = "ZANJÓN"


class TipoCaudal(str, Enum):
    PERMANENTE = "PERMANENTE"
    EVENTUAL = "EVENTUAL"
    CAUDAL_MAXIMO_INSTANTANEO = "CAUDAL MÁXIMO INSTANTÁNEO"
    NO_INDICA = "NO INDICA/OTRO"

class UnidadCaudal(str, Enum):
    LITROS_POR_MINUTO = "L/MIN"
    LITROS_POR_SEGUNDO = "L/S"
    LITROS_POR_HORA = "L/H"
    LITROS_POR_DIA = "L/DIA"
    LITROS_POR_MES = "L/MES"
    METROS_CUBICOS_POR_SEGUNDO = "M3/S"
    METROS_CUBICOS_POR_MINUTO = "M3/MIN"
    METROS_CUBICOS_POR_HORA = "M3/H"
    METROS_CUBICOS_POR_DIA = "M3/DIA"
    METROS_CUBICOS_POR_MES = "M3/MES"
    METROS_CUBICOS_ANUAL = "M3/ANUAL"
    METROS_CUBICOS_POR_HA = "M3/HA ANUALES"
    ACCIONES = "ACCIONES"
    REGALIAS = "REGALÍAS"
    REGADORES = "REGADORES"
    HORAS_TURNO = "HORAS TURNO"
    PARTES = "PARTES"
    CUOTA = "CUOTA"
    CUADRA = "CUADRA"
    TEJA = "TEJA"
    PORCENTAJE = "%"
    NO_INDICA = "NO INDICA/OTRO"


PositiveFloat = Annotated[float, Field(ge=0)]

class Fase4Registro(BaseModel):
    tipo_caudal: List[TipoCaudal] = Field(description=descr["tipo_caudal"])
    unidad_caudal: UnidadCaudal = Field(description=descr["unidad_caudal"])
    caudal_promedio: Optional[List[PositiveFloat]] = Field(None, description=descr["caudal_promedio"])
    tipo_fuente: Optional[List[TipoFuentes]] = Field(None, description=descr["tipo_fuente"])
    fuente: Optional[List[str]] = Field(None, description=descr["fuente"])
    cuenca: Optional[List[str]] = Field(None, description=descr["cuenca"])  

for mes in ['ene', 'feb', 'mar', 'abr', 'may', 'jun', 'jul', 'ago', 'sep', 'oct', 'nov', 'dic']:
    campo = f'caudal_{mes}'
    Fase4Registro.__annotations__[campo] = Optional[List[Optional[PositiveFloat]]]
    setattr(Fase4Registro, campo, Field(None, description=f"Caudal para el mes de {mes}. Si un derecho no tiene detalle mensual, usar null."))

Fase4Registro.model_rebuild()
```
